## Packages

```python
import numpy as np
import pandas as pd
import os
from pyprojroot import here
```

## Load data

```python
df = pd.read_csv(os.path.join(here(), "data", "vowel_duration_toy_data.csv"))
df.head()
```

## Print a summary table

```python
# Quick summary statistics
print(df.shape)
print(df.dtypes)
print()
```

```python

# Overall numeric summary
print(df.describe())
print()
```

```python

# Grouped summary by condition
summary = (
    df.groupby("condition")["vowel_duration_ms"]
    .agg(["count", "mean", "std", "min", "max"])
    .round(2)
)
print(summary)
```
