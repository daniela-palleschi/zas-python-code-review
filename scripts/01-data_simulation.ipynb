```python

# Reproducibility
np.random.seed(416)
```

```python

# Simulation parameters
n_speakers = 20
n_words_per_condition = 15
conditions = ["voiced", "voiceless"]
```

```python

# True effect sizes (in milliseconds)
base_vowel_duration = 120   # baseline duration in ms
voicing_effect = 35         # voiced consonants add ~35ms to vowel duration
speaker_sd = 15             # variability between speakers
trial_sd = 12               # trial level noise
```

```python

rows = []

for speaker_id in range(1, n_speakers + 1):
    # Each speaker has their own baseline tendency
    speaker_offset = np.random.normal(0, speaker_sd)

    for condition in conditions:
        effect = voicing_effect if condition == "voiced" else 0

        for trial in range(1, n_words_per_condition + 1):
            duration = (
                base_vowel_duration
                + effect
                + speaker_offset
                + np.random.normal(0, trial_sd)
            )

            rows.append({
                "speaker_id": f"S{speaker_id:02d}",
                "condition": condition,
                "trial": trial,
                "vowel_duration_ms": round(duration, 1)
            })

df = pd.DataFrame(rows)
```

```python

# Quick summary
summary = df.groupby("condition")["vowel_duration_ms"].agg(["mean", "std", "count"])
print(summary)
```

```python

# Save to CSV
df.to_csv(os.path.join(here(), "data", "vowel_duration_toy_data.csv"), index=False)
print("\nSaved to vowel_duration_toy_data.csv")
```
